### Why 700 pounds?

The tongue‑twister “How much wood would a woodchuck chuck?” is mostly a play on words, but it has inspired a little bit of science. In 1988, wildlife biologist Richard Thomas ran a rough calculation:

| Item | Weight per unit | Units in a burrow | Total weight |
|------|-----------------|-------------------|--------------|
| Woodchuck body (average 5 lb) | 5 lb | 1 | 5 lb |
| Soil displaced (average 6 lb/ft³) | 6 lb | 15 ft³ (a typical burrow) | ~90 lb |
| Wood moved (assuming a 1‑ft³ volume, ~35 lb/ft³) | 35 lb | 5 ft³ (the amount a chuck might move in a day) | ~175 lb |

Thomas then multiplied the daily volume by an estimate of how many hours a woodchuck could “chuck” in a day, arriving at roughly **700 pounds (≈320 kg)** of wood over the course of a year.

> **Note:** The calculation is highly speculative—woodchucks (aka groundhogs) are burrowing animals that move dirt, not logs. The figure is more a fun illustration than a hard science fact.

### Bottom line

```python
import asyncio
from together import AsyncTogether

async def send_request(client, idx):
    try:
        response = await client.chat.completions.create(
            model=model_endpoint,
            messages=[
                {
                    "role": "user",
                    "content": f"How much wood could a wood chuck chuck if a wood chuck could chuck wood? (Request {idx})"
                }
            ]
        )
        print(f"Response {idx}: {response.choices[0].message.content[:60]}...")
    except Exception as e:
        print(f"Request {idx} failed: {e}")

async def main():
    client = AsyncTogether()
    tasks = [send_request(client, i,) for i in range(1, 25)]
    await asyncio.gather(*tasks)

# Run the async main function
await main()
```
